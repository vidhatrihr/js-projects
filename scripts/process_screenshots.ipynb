```python
from PIL import Image
import os
import json
```

```python
img_path = 'joachim-lesne-couTAixLzNM-unsplash.jpg'

img = Image.open(img_path)


def crop_to_aspect(img, aspect_ratio=1.33):
  width, height = img.size
  new_height = round(width / aspect_ratio)
  if height <= new_height:
    return img, False

  return img.crop((0, 0, width, new_height))
```

```python
def resize_width(img, new_width=500):
  width, height = img.size
  new_height = round((new_width * height) / width)
  if width <= new_width:
    return img, False

  return img.resize((new_width, new_height))
```

```python
f = open('projects_data.json')
data = json.load(f)
```

```python
def process_screenshots():
  max_width = 400
  for project in data:
    img_path = f'./assets/screenshots/{project['slug']}.jpg'
    img = Image.open(img_path)

    width, height = img.size

    # crop img
    img, cropped = crop_to_aspect(img)

    # resize img
    img, resized = resize_width(img, max_width)

    # convert to RGB
    converted = False
    if img.mode in ('RGBA', 'LA'):
      img = img.convert('RGB')
      converted = True

    output_path = os.path.join(os.path.dirname(img_path), f'{project['slug']}.jpg')
    img.save(output_path, 'JPEG', quality=95)

    status = []
    status.append("✅ cropped" if cropped else "❌ skipped cropping")
    status.append("✅ resized" if resized else "❌ skipped resizing")
    if converted:
      status.append('converted to RGB')

    print(f'Processed {project['slug']}.jpg ({", ".join(status)})')
```

```python
process_screenshots()
```
